**Calculator Program (Add, Subtract, Multiply, Divide, Power)**

```python
# Simple Calculator

print("Simple Calculator")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")
print("5. Power")

choice = input("Enter choice (1-5): ")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if choice == '1':
    print("Result:", num1 + num2)

elif choice == '2':
    print("Result:", num1 - num2)

elif choice == '3':
    print("Result:", num1 * num2)

elif choice == '4':
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Division by zero not allowed!")

elif choice == '5':
    print("Result:", num1 ** num2)

else:
    print("Invalid Choice")
```

**Check if List Contains Any Numeric Value (Using for loop)**

```python
my_list = ["Bilal", "Python", 25, "Code", 3.14]

found = False

for item in my_list:
    if isinstance(item, (int, float)):
        found = True
        break

if found:
    print("List contains numeric value.")
else:
    print("No numeric value found.")
```

**Add a Key to Dictionary**

```python
my_dict = {"name": "Bilal", "age": 25}

# Adding new key
my_dict["city"] = "Karachi"

print(my_dict)
```

**Sum All Numeric Items in Dictionary**

```python
my_dict = {"a": 10, "b": 20, "c": "Python", "d": 5}

total = 0

for value in my_dict.values():
    if isinstance(value, (int, float)):
        total += value

print("Sum of numeric values:", total)
```

**Identify Duplicate Values from List**

```python
my_list = [1, 2, 3, 4, 2, 5, 1, 6]

duplicates = []

for item in my_list:
    if my_list.count(item) > 1 and item not in duplicates:
        duplicates.append(item)

print("Duplicate values:", duplicates)
```

**Check if Given Key Exists in Dictionary**
