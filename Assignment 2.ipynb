```python
#1 Print the length of a list
my_list = [10, 20, 30, 40, 50]

length = len(my_list)

print("Length of the list is:", length)
```

```python
#2 Sum all numeric items in a list
my_list = [10, 20, 30, 40, 50]

total = sum(my_list)

print("Sum of the list is:", total)
```

```python
#3 Get the largest number from a numeric list
my_list = [10, 45, 67, 23, 89, 12]

largest = max(my_list)

print("Largest number in the list is:", largest)
```

```python
#4 Print elements less than 5
a = [1, 2, 3, 4, 5, 8, 13, 21, 34, 55, 89]

for item in a:
    if item < 5:
        print(item)
```

```python
#5 Alternative (Short Method for Task 4)
a = [1, 2, 3, 4, 5, 8, 13, 21, 34, 55, 89]

result = [x for x in a if x < 5]

print(result)
```
